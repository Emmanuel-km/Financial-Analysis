```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Define Forecast Data
months = [
    'Jan-27',
    'Feb-27',
    'Mar-27',
    'Apr-27',
    'May-27',
    'Jun-27',
    'Jul-27',
    'Aug-27',
    'Sep-27',
    'Oct-27',
    'Nov-27',
    'Dec-27',
]

naive = [46.79] * 12
rw_drift = [
    47.52,
    48.27,
    49.03,
    49.79,
    50.58,
    51.37,
    52.17,
    52.99,
    53.82,
    54.67,
    55.53,
    56.40,
]
damped_holt = [
    47.30,
    47.71,
    48.09,
    48.44,
    48.77,
    49.08,
    49.36,
    49.62,
    49.86,
    50.09,
    50.29,
    50.49,
]
arima = [
    47.11,
    47.75,
    48.46,
    49.20,
    49.94,
    50.70,
    51.47,
    52.26,
    53.05,
    53.86,
    54.68,
    55.51,
]
linear_trend = [
    51.81,
    53.21,
    54.60,
    56.00,
    57.40,
    58.80,
    60.19,
    61.59,
    62.99,
    64.38,
    65.78,
    67.18,
]

# 2. Plot Configuration (Dark Theme matching your visual)
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(12, 6), dpi=150)

# Plotting Forecast Scenarios
ax.plot(
    months,
    naive,
    label='Pessimistic Scenario (Naive Baseline: 46.79)',
    color='#FF4B4B',
    linestyle=':',
    marker='o',
    linewidth=2,
)
ax.plot(
    months,
    damped_holt,
    label='Conservative Growth (Damped Holt)',
    color='#36A2EB',
    linestyle=':',
    marker='s',
    linewidth=1.8,
)
ax.plot(
    months,
    arima,
    label='Base Case (ARIMA 1,1,0 + Drift)',
    color='#FFCE56',
    linestyle=':',
    marker='^',
    linewidth=2.2,
)
ax.plot(
    months,
    linear_trend,
    label='Optimistic Scenario (Linear Trend 24M: 67.18)',
    color='#2ECC71',
    linestyle=':',
    marker='D',
    linewidth=2.2,
)

# 3. Aesthetics & Annotations
ax.set_title(
    'TOTL Stock Price Forecast Outlook (2027)\nRock Data Plus Predictive Modeling',
    fontsize=14,
    fontweight='bold',
    pad=15,
)
ax.set_ylabel('Stock Price (KES)', fontsize=11)
ax.set_xlabel('Forecast Horizon (2027)', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.3)
ax.set_ylim(40, 72)

# End-point Annotations
ax.annotate(
    f'Pessimistic: {naive[-1]}',
    xy=(11, naive[-1]),
    xytext=(11.2, naive[-1]),
    color='#FF4B4B',
    fontweight='bold',
)
ax.annotate(
    f'Base (ARIMA): {arima[-1]}',
    xy=(11, arima[-1]),
    xytext=(11.2, arima[-1]),
    color='#FFCE56',
    fontweight='bold',
)
ax.annotate(
    f'Optimistic: {linear_trend[-1]}',
    xy=(11, linear_trend[-1]),
    xytext=(11.2, linear_trend[-1]),
    color='#2ECC71',
    fontweight='bold',
)

plt.xticks(rotation=45)
plt.legend(loc='upper left', frameon=True, facecolor='#1E1E1E')
plt.tight_layout()
plt.show()
```
